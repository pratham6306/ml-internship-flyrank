```python
!ls ml-internship-flyrank/data/raw
import pandas as pd

df = pd.read_csv("/content/ml-internship-flyrank/data/raw/content_refresh_anonymized.csv")

print(f"Rows: {len(df)}")
print(f"Columns: {len(df.columns)}")

corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d: {corr:.3f}")

print("\nMedian Impressions:", df["impressions_90d"].median())
print("Median Sessions:", df["sessions_90d"].median())
```

### Quick Look at the Data

The starter dataset contains **30,000 webpages** described by **44 features**, providing enough observations to explore patterns related to content performance.

The correlation between **search_volume** and **impressions_90d** is **0.001**, indicating that keyword search volume alone does not explain how much traffic a page receives. This suggests that prioritizing pages for review requires combining multiple observable signals rather than relying on a single metric.

The median webpage received **731 impressions** but only **7 sessions** over the previous 90 days. This highlights that visibility does not always translate into user engagement, making content prioritization a meaningful decision-support problem.
